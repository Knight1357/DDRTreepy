```python
import numpy as np
import pandas as pd


def generate_test_data():

    # 设置随机种子
    np.random.seed(42)

    # 参数设置
    n_samples = 600      # 样本数量（N）
    n_features = 500    # 原始特征维度（D）
    num_clusters = 100   # 聚类数量（K）
    dimensions = 2      # 降维后的维度（d）
    maxiter = 20        # 最大迭代次数
    sigma = 1e-3        # 高斯核参数
    lambda_ = 0.1       # 正则化参数
    gamma = 10         # 权重参数
    eps = 1e-3          # 收敛阈值
    verbose = True      # 是否输出详细信息

    # 生成可控的随机数据
    R_X = np.random.rand(n_features, n_samples)          # (D x N)
    R_Z = np.random.rand(dimensions, n_samples)          # (d x N)
    R_Y = np.random.rand(dimensions, num_clusters)       # (d x K)
    R_W = np.random.rand(n_features, dimensions)         # (D x d)

    # 检查输入数据是否有异常值
    for name, matrix in zip(['R_X', 'R_Z', 'R_Y', 'R_W'], [R_X, R_Z, R_Y, R_W]):
        assert not np.any(np.isnan(matrix)), f"{name} contains NaN values"
        assert not np.any(np.isinf(matrix)), f"{name} contains Inf values"

    R_X_path = '../data/R_X.csv'
    R_Z_path = '../data/R_Z.csv'
    R_Y_path = '../data/R_Y.csv'
    R_W_path = '../data/R_W.csv'
    
    df = pd.DataFrame(R_X)
    df.to_csv(R_X_path, index=False, header=False)  
    print(f"随机矩阵已保存为 {R_X_path}")
    
    df = pd.DataFrame(R_Z)
    df.to_csv(R_Z_path, index=False, header=False)  
    print(f"随机矩阵已保存为 {R_Z_path}")
    
    df = pd.DataFrame(R_Y)
    df.to_csv(R_Y_path, index=False, header=False)  
    print(f"随机矩阵已保存为 {R_Y_path}")
    
    df = pd.DataFrame(R_W)
    df.to_csv(R_W_path, index=False, header=False)  
    print(f"随机矩阵已保存为 {R_W_path}")

if __name__ == "__main__":
    generate_test_data()
```
